# Plots

**Common values**

```python
dataset = str(ff.EUMACO_WT)
dataset_short = str(ff.EUMACO_WT_s)
# figsize = (8, 6)
progr = 0
output_format = 'png'
savefig = False
```

## Plot for wt field

```python
figsize = (8, 6)
y_label = 'Count'
```

### `tran.timestamp` - hour distribution

```python
colname = 'tran.timestamp'
title = f'{dataset_short} Transmissions counts by hour of day'
x_label = 'Hour'
# y_label = ''
output_name = 'timestamp_hour'
progr += 1
```

```python
fn.bar_plot(df[colname].dt.hour.value_counts().reindex(range(24), fill_value=0),
    title, x_label, y_label,
    dataset, progr, output_name, output_format,
    kind='time', label_limit=100,
    savefig=savefig)
```

### `tran.timestamp` - date distribution

```python
colname = 'tran.timestamp'
title = f'{dataset_short} Transmissions counts by day'
x_label = 'Year'
# y_label = ''
output_name = 'timestamp_date'
progr += 1
```

```python
fn.bar_plot(df[colname].dt.year.value_counts().sort_index(),
    title, x_label, y_label,
    dataset, progr, output_name, output_format,
    kind='date', label_limit=500,
    savefig=savefig)
```

### `tran.weig.value`

```python
colname = 'tran.weig.value'
title = 'Weight distribution'
x_label = 'kg'
# y_label = ''
output_name = 'weight'
progr += 1
```

```python
fn.histbox_plot(df[colname],
    title, x_label, y_label,
    dataset, progr, output_name, output_format,
    figsize, bins='auto',
    savefig=savefig)
```
