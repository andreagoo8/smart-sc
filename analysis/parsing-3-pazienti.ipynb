## Analysis

### Cross-analysis of pseudopazs from other datasets

```python
folder_path = ff.PROCESSED_DATA_DIR / ff.PICKLES / ff.PSEUDOPATIENTS / '*-list.pickle'

path_list = sorted(glob.glob(str(folder_path)))

# Drop from list `3-pazienti` (this dataset pps)
path_list.remove(glob.glob(str(ff.PROCESSED_DATA_DIR / ff.PICKLES / ff.PSEUDOPATIENTS / '3-*-list.pickle'))[0])
```

#### For `fornitore` column

```python
for path in path_list:
    with open(path, 'rb') as file:
        other_ppazs = pickle.load(file)

    wdf = df[df['pseudopaz'].isin(other_ppazs)]
    print(path[path.rfind('/')+1 : path.rfind('-pp-uniq-list.pickle')])
    print(f'\t{wdf['fornitore'].value_counts(dropna=False).to_dict()}\n')
```

#### For `tipo` column

```python
for path in path_list:
    with open(path, 'rb') as file:
        other_ppazs = pickle.load(file)

    wdf = df[df['pseudopaz'].isin(other_ppazs)]
    print(path[path.rfind('/')+1 : path.rfind('-pp-uniq-list.pickle')])
    print(f'\t{wdf['tipo'].value_counts(dropna=False).to_dict()}\n')
```
